```python

# EXPERIMENT 15: Data Preprocessing in Data Science

import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, MinMaxScaler
from sklearn.compose import ColumnTransformer

# Step 1: Create sample dataset with missing values
data = {
    "Country": [
        "France", "Spain", "Germany", "Spain", "Germany",
        "France", "Spain", "France", None, "France"
    ],
    "Age": [44, 27, 30, 38, 40, 35, np.nan, 48, 50, 37],
    "Salary": [
        72000, 48000, 54000, 61000, np.nan,
        58000, 52000, 79000, 83000, 67000
    ],
    "Purchased": [
        "No", "Yes", "No", "No", "Yes",
        "Yes", "No", "Yes", "No", "Yes"
    ]
}

df = pd.DataFrame(data)

# Step 2: Save the dataset as CSV
df.to_csv("pre-process_datasample.csv", index=False)
print("CSV file created successfully!")

# Step 3: Read the CSV file
df = pd.read_csv("pre-process_datasample.csv")

print("\nOriginal Dataset:")
display(df)

# Step 4: Display dataset information
print("\nDataset Information:")
df.info()

# Step 5: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 6: Fill missing Country values using mode
df["Country"] = df["Country"].fillna(df["Country"].mode()[0])

# Step 7: Fill missing Age values using median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Step 8: Fill missing Salary values using mean
df["Salary"] = df["Salary"].fillna(round(df["Salary"].mean()))

print("\nDataset after handling missing values:")
display(df)

# Step 9: Encode categorical data using One-Hot Encoding
df_encoded = pd.get_dummies(
    df,
    columns=["Country"],
    dtype=int
)

# Step 10: Encode Purchased column
df_encoded["Purchased"] = df_encoded["Purchased"].map(
    {"No": 0, "Yes": 1}
)

print("\nDataset after encoding categorical variables:")
display(df_encoded)

# Step 11: Standardize numerical features
standard_scaler = StandardScaler()

df_encoded[["Age", "Salary"]] = standard_scaler.fit_transform(
    df_encoded[["Age", "Salary"]]
)

print("\nDataset after Standard Scaling:")
display(df_encoded)

# Step 12: Apply Min-Max Scaling to numerical features
minmax_scaler = MinMaxScaler()

df_minmax = df.copy()
df_minmax[["Age", "Salary"]] = minmax_scaler.fit_transform(
    df_minmax[["Age", "Salary"]]
)

print("\nDataset after Min-Max Scaling:")
display(df_minmax)

# Step 13: Save the preprocessed dataset
df_encoded.to_csv("preprocessed_data.csv", index=False)

print("\nExperiment 15 completed successfully!")
print("Output saved as preprocessed_data.csv")
```
