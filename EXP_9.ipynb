```python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from google.colab import files

# Display plots inline
%matplotlib inline

# Step 1: Upload and Load Dataset
# If file is not present locally in Colab, prompt to upload
if not os.path.exists('Iris_Dataset.csv'):
    print("Please upload Iris_Dataset.csv:")
    uploaded = files.upload()

# Load the Iris dataset
data = pd.read_csv('Iris_Dataset.csv')

# Step 2: Display Basic Information & Summary Statistics
print("First 5 rows:")
print(data.head())

print("\nDataset Information:")
data.info()

print("\nSummary Statistics:")
print(data.describe())

# Count the number of samples in each variety
print("\nVariety Counts:")
print(data['variety'].value_counts())

# Step 3: Visualizations
# Count plot of Iris varieties
plt.figure(figsize=(7, 5))
sns.countplot(x='variety', data=data)
plt.title('Count of Iris Varieties')
plt.xlabel('Variety')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

# One-hot encoding of the variety column
dummies = pd.get_dummies(data['variety'], dtype=int)

# Combine encoded columns with numerical features
FinalDataset = pd.concat([dummies, data.iloc[:, :4]], axis=1)
print("\nFinal Dataset:")
print(FinalDataset.head())

# Scatter plot: Sepal length vs Sepal width
plt.figure(figsize=(7, 5))
sns.scatterplot(x='SepalLengthCm', y='SepalWidthCm', hue='variety', data=data)
plt.title('Sepal Length vs Sepal Width')
plt.tight_layout()
plt.show()

# Scatter plot: Petal length vs Petal width
plt.figure(figsize=(7, 5))
sns.scatterplot(x='PetalLengthCm', y='PetalWidthCm', hue='variety', data=data)
plt.title('Petal Length vs Petal Width')
plt.tight_layout()
plt.show()

# Pair plot of all numerical features
sns.pairplot(data, hue='variety', height=3)
plt.show()

# Histograms for each feature by variety
features = ['PetalLengthCm', 'PetalWidthCm', 'SepalLengthCm', 'SepalWidthCm']
for feature in features:
    g = sns.FacetGrid(data, hue='variety', height=5)
    g.map(sns.histplot, feature, kde=False)
    g.add_legend()
    g.fig.suptitle(f'Distribution of {feature}', y=1.02)
    plt.show()
```
