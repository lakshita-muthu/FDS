```python

# EXPERIMENT 11: Pandas DataFrame Functions

import pandas as pd
import numpy as np

# Step 1: Create sample employee data
employees = {
    "EmpID": [1, 2, 3, 4, 5],
    "Name": ["Smith", "Jones", "Taylor", "Patel", "Kumar"],
    "Age": [25, 30, 28, 35, 32],
    "Salary": [50000, 60000, 55000, 70000, 65000]
}

df = pd.DataFrame(employees)

# Step 2: Save data to CSV
df.to_csv("employees.csv", index=False)
print("employees.csv created successfully!")

# Step 3: Read CSV file
df = pd.read_csv("employees.csv")

# Step 4: Display the DataFrame
print("\nEmployee DataFrame:")
display(df)

# Step 5: Display information
print("\nDataFrame Information:")
df.info()

# Step 6: Display first and last rows
print("\nFirst five rows:")
display(df.head())

print("\nLast five rows:")
display(df.tail())

# Step 7: Display number of rows and columns
print("\nShape:", df.shape)

# Step 8: Select a single column
print("\nEmployee Names:")
print(df["Name"])

# Step 9: Select multiple columns
print("\nName and Salary:")
display(df[["Name", "Salary"]])

# Step 10: Select rows using a condition
print("\nEmployees with Salary greater than 55000:")
display(df[df["Salary"] > 55000])

# Step 11: Add a new column
df["Department"] = [
    "IT", "HR", "Sales", "IT", "Finance"
]
print("\nAfter adding Department:")
display(df)

# Step 12: Rename a column
df = df.rename(columns={"Name": "EmployeeName"})
print("\nAfter renaming Name:")
display(df)

# Step 13: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 14: Sort by salary
print("\nSorted by Salary:")
display(df.sort_values(by="Salary"))

# Step 15: Group employees by department
print("\nEmployees in Each Department:")
print(df.groupby("Department").size())

# Step 16: Calculate statistical measures
print("\nMean Salary:", df["Salary"].mean())
print("Median Salary:", df["Salary"].median())
print("Maximum Salary:", df["Salary"].max())
print("Minimum Salary:", df["Salary"].min())
print("Standard Deviation:", df["Salary"].std())

# Step 17: Display descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe())

# Step 18: Convert DataFrame to NumPy array
print("\nDataFrame as NumPy Array:")
print(df.to_numpy())

# Step 19: Save the final DataFrame
df.to_csv("employees_processed.csv", index=False)

print("\nExperiment 11 completed successfully!")
```
