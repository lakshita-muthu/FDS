```python

import numpy as np

# Create a random array of 9 integers between 1 and 99
array = np.random.randint(1, 100, 9)
print("Original Array:")
print(array)

# Calculate the square root of each element
print("\nSquare Root:")
print(np.sqrt(array))

# Check the number of dimensions
print("\nOriginal Array Dimensions:", array.ndim)

# Reshape the array into a 3x3 matrix
new_array = array.reshape(3, 3)
print("\nReshaped 3x3 Array:")
print(new_array)

# Check the dimensions of the reshaped array
print("\nReshaped Array Dimensions:", new_array.ndim)

# Flatten the array into a 1D array
print("\nFlattened Array:")
print(new_array.ravel())

# Reshape into a 3x3 matrix again
newm = new_array.reshape(3, 3)
print("\nNew Matrix:")
print(newm)

# Array slicing examples
print("\nElements in row 3, columns 2 to 3:")
print(newm[2, 1:3])

print("\nRow 2, columns 2 to 3:")
print(newm[1:2, 1:3])

print("\nAll rows, no columns (empty array):")
print(new_array[0:3, 0:0])

print("\nFirst 2 rows, first column:")
print(new_array[0:2, 0:1])

print("\nAll rows, first column:")
print(new_array[0:3, 0:1])

print("\nRows 2 and 3:")
print(new_array[1:3])
```
